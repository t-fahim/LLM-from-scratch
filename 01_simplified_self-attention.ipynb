```python
import torch

inputs = torch.tensor(
    [[0.72, 0.45, 0.31], # Dream  (x^1)
     [0.75, 0.20, 0.55], # big    (x^2)
     [0.30, 0.80, 0.40], # and    (x^3)
     [0.85, 0.35, 0.60], # work   (x^4)
     [0.55, 0.15, 0.75], # for    (x^5)
     [0.25, 0.20, 0.85]] # it     (x^6)
)

# Corresponding words
words = ['Dream', 'big', 'and', 'work', 'for', 'it']
```

```python
# Calculate the magnitude of each vector
magnitudes = torch.norm(inputs, dim=1)

# Print the magnitudes
print("Magnitudes of the vectors:")
for word, magnitude in zip(words, magnitudes):
    print(f"{word}: {magnitude.item():.4f}")
```

```python
import matplotlib.pyplot as plt

# Extract x, y, z coordinates
x_coords = inputs[:, 0].numpy()
y_coords = inputs[:, 1].numpy()
z_coords = inputs[:, 2].numpy()

# 3D plot with vectors from origin
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

colors = ['r', 'g', 'b', 'c', 'm', 'y']

for (x, y, z, word, color) in zip(x_coords, y_coords, z_coords, words, colors):
    ax.quiver(0, 0, 0, x, y, z, color=color, arrow_length_ratio=0.05)
    ax.text(x, y, z, word, fontsize=10, color=color)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_xlim([0, 1])
ax.set_ylim([0, 1])
ax.set_zlim([0, 1])
plt.title('3D Plot of Word Embeddings with Colored Vectors')
plt.show()
```

## Dot product beteen 2nd input token and all ohter token

```python
query = inputs[1]  # 2nd input token is the query

attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)  # dot product (transpose not necessary here)

print(attn_scores_2)
```

## Normalizing the attension score

```python
attn_weights_2_tmp = attn_scores_2 / attn_scores_2.sum()

print("Attention weights:", attn_weights_2_tmp)
print("Sum:", attn_weights_2_tmp.sum())
```

## Normalizing usnig Sofrmax

```python
# Compute attention weights using PyTorch's softmax function
attn_weights_2_tmp = torch.softmax(attn_scores_2, dim=0)

print("Attention weights:", attn_weights_2_tmp)
print("Sum:", attn_weights_2_tmp.sum().item())
```

## Attension score for all Querie

```python
# Calculate attention scores for all query-key pairs using matrix multiplication
attn_scores = torch.matmul(inputs, inputs.T)

# Print the resulting 6x6 attention scores matrix
print(attn_scores)
```

## Attnsion score is convertend into attension weight (softmax)

```python
# Convert attention scores to attention weights using softmax along the last dimension
attn_weights = torch.softmax(attn_scores, dim=-1)

# Print the resulting attention weights matrix
print(attn_weights)
```

```python
# Compute context vectors for all queries via matrix multiplication
all_context_vecs = torch.matmul(attn_weights,inputs)

# Print the resulting context vectors matrix
print(all_context_vecs)
```

```python
# Context vector corresponding to the 2nd input
context_vec_2 = all_context_vecs[1]
print("Previous 2nd context vector:", context_vec_2)
```

```python
# Adding context vector just for plotting

# Append context_vec_2 to inputs
inputs = torch.cat((inputs, context_vec_2.unsqueeze(0)), dim=0)

# Add 'context_vector' to the words list
words.append('context_vector_for_big')

print("Updated inputs tensor:")
print(inputs)
```

```python
import matplotlib.pyplot as plt

# Extract x, y, z coordinates
x_coords = inputs[:, 0].numpy()
y_coords = inputs[:, 1].numpy()
z_coords = inputs[:, 2].numpy()

# 3D plot with vectors from origin
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

colors = ['orange', 'g', 'b', 'c', 'm', 'y','r']

for (x, y, z, word, color) in zip(x_coords, y_coords, z_coords, words, colors):
    ax.quiver(0, 0, 0, x, y, z, color=color, arrow_length_ratio=0.05)
    ax.text(x, y, z, word, fontsize=10, color=color)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_xlim([0, 1])
ax.set_ylim([0, 1])
ax.set_zlim([0, 1])
plt.title('3D Plot of Word Embeddings with Colored Vectors')
plt.show()
```
